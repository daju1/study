```python
from IPython.display import display, Math, Latex

def disp(ex):
    display(Math(latex(ex)))

def displ(tex):
    display(Math(tex))
```

```python
a,r,h = var("a,r,h")
gamma_1, gamma_2 = var("gamma_1, gamma_2")
alpha = var("alpha")
k, l_0 = var("k, l_0")
```

```python
l_1 = sqrt(a^2+r^2-2*r*a*cos(gamma_1))
```

```python
l_2 = sqrt(a^2+r^2-2*r*a*cos(gamma_2))
```

```python
eq_gamma = gamma_1 + gamma_2 + 2*alpha == pi
eq_alpha = alpha == acos(h/a)
```

```python
x_1 = l_1 - l_0
x_2 = l_2 - l_0
```

```python
U = k / 2 * (x_1^2 + x_2^2) + k / 2 * (x_1^3 + x_2^3)
disp(U)
```

```python
sol_gamma_2 = solve(eq_gamma, gamma_2)
disp(sol_gamma_2)
```

```python
gamma_1_degree = var("gamma_1_degree")
eq_degree = gamma_1 == gamma_1_degree / 180 * pi
U = U.subs(sol_gamma_2).subs(eq_alpha).subs(eq_degree)
disp(U)
```

```python
values = [
    k == 1,
    a == 10,
    r == 1,
    l_0 == 9.25,
    h == 9.5,
]
```

```python
disp(U.subs(values))
```

```python
plot([U.subs(values), 0], gamma_1_degree, 0, 720)
```

```python
import curses
from curses.textpad import Textbox, rectangle

def main(stdscr):
    stdscr.addstr(0, 0, "Enter IM message: (hit Ctrl-G to send)")

    editwin = curses.newwin(5,30, 2,1)
    rectangle(stdscr, 1,0, 1+5+1, 1+30+1)
    stdscr.refresh()

    box = Textbox(editwin)

    # Let the user edit until Ctrl-G is struck.
    box.edit()

    # Get resulting contents
    message = box.gather()
```

```python
from curses import wrapper

def main(stdscr):
    # Clear screen
    stdscr.clear()

    # This raises ZeroDivisionError when i == 10.
    for i in range(0, 11):
        v = i-10
        stdscr.addstr(i, 0, '10 divided by {} is {}'.format(v, 10/v))

    stdscr.refresh()
    stdscr.getkey()

wrapper(main)
```
